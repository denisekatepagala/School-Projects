## Import libraries

```python
import tensorflow as tf 
import matplotlib.pyplot as plt 
import pandas as pd 
import seaborn as sns 
```

## Data loading and Image Preprocessing

```python
training_set = tf.keras.utils.image_dataset_from_directory(
    './Plant_Disease_Dataset/train',
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(256, 256),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
)
```

```python
validation_set = tf.keras.utils.image_dataset_from_directory(
    './Plant_Disease_Dataset/valid',
    labels="inferred",
    label_mode="categorical",
    class_names=None,
    color_mode="rgb",
    batch_size=32,
    image_size=(256, 256),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation="bilinear",
    follow_links=False,
    crop_to_aspect_ratio=False,
)
```

## Building the CNN Model

```python
training_set
```

```python
for x,y in training_set:
    print(x,x.shape)
    print(y,y.shape)
    break
```

```python
from tensorflow.keras.layers import Dense,Conv2D,MaxPool2D,Flatten,Dropout
from tensorflow.keras.models import Sequential
```

```python
cnn = Sequential()
```

```python
cnn.add(Conv2D(filters=32,kernel_size=3,padding='same',activation='relu',input_shape=[256,256,3]))
cnn.add(Conv2D(filters=32,kernel_size=3,activation='relu'))
cnn.add(MaxPool2D(pool_size=2,strides=2))
```

```python
cnn.add(Conv2D(filters=64,kernel_size=3,padding='same',activation='relu'))
cnn.add(Conv2D(filters=64,kernel_size=3,activation='relu'))
cnn.add(MaxPool2D(pool_size=2,strides=2))
```

```python
cnn.add(Conv2D(filters=128,kernel_size=3,padding='same',activation='relu'))
cnn.add(Conv2D(filters=128,kernel_size=3,activation='relu'))
cnn.add(MaxPool2D(pool_size=2,strides=2))
```

```python
cnn.add(Conv2D(filters=256,kernel_size=3,padding='same',activation='relu'))
cnn.add(Conv2D(filters=256,kernel_size=3,activation='relu'))
cnn.add(MaxPool2D(pool_size=2,strides=2))
```

```python
cnn.add(Conv2D(filters=512,kernel_size=3,padding='same',activation='relu'))
cnn.add(Conv2D(filters=512,kernel_size=3,activation='relu'))
cnn.add(MaxPool2D(pool_size=2,strides=2))
```

```python
cnn.add(Flatten())
cnn.add(Dense(units=512, activation='relu'))
cnn.add(Dropout(0.5))
cnn.add(Dense(units=38, activation='softmax'))
```

```python
cnn.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001), # Standard string 'adam' is 0.001 (too high for this depth)
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
cnn.summary()
```

## Model Training and Evaluation

```python
training = cnn.fit(x=training_set, validation_data=validation_set, epochs=20)
```

```python
train_loss,train_acc = cnn.evaluate(training_set)
print(train_loss,train_acc)
```

```python
val_loss,val_acc = cnn.evaluate(validation_set)
print(val_loss,val_acc)
```
